```python
# Data awal hasil panen
data_panen = [
    {'id': 'P001', 'komoditas': 'Padi', 'stok': 500, 'harga': 7000},
    {'id': 'P002', 'komoditas': 'Jagung', 'stok': 300, 'harga': 5000},
    {'id': 'P003', 'komoditas': 'Kedelai', 'stok': 200, 'harga': 9000},
    {'id': 'P004', 'komoditas': 'Cabai', 'stok': 100, 'harga': 35000}
]

# Riwayat transaksi
riwayat_transaksi = []

# Total pendapatan
total_pendapatan = 0


def tampilkan_data_panen():
    """Menampilkan data komoditas panen, stok, dan harga per kg."""
    print("\n--- Data Panen Tersedia ---")

    if not data_panen:
        print("Tidak ada data panen saat ini.")
        return

    print(f"{'ID':<5} {'Komoditas':<15} {'Stok (kg)':<10} {'Harga/kg':<15}")
    print("-" * 45)

    for item in data_panen:
        print(
            f"{item['id']:<5} "
            f"{item['komoditas']:<15} "
            f"{item['stok']:<10d} "
            f"{'Rp' + f'{item['harga']:,.0f}':<14}"
        )

    print()


def hitung_diskon(total_harga):
    """
    Menghitung diskon berdasarkan total harga pembelian.

    Kurang dari Rp500.000      : 0%
    Rp500.000 - Rp999.999     : 5%
    Rp1.000.000 atau lebih    : 10%
    """

    if total_harga >= 1_000_000:
        persentase_diskon = 10
    elif total_harga >= 500_000:
        persentase_diskon = 5
    else:
        persentase_diskon = 0

    jumlah_diskon = total_harga * persentase_diskon / 100
    total_setelah_diskon = total_harga - jumlah_diskon

    return persentase_diskon, jumlah_diskon, total_setelah_diskon


def jual_panen():
    """
    Melakukan penjualan hasil panen, menghitung diskon,
    mengurangi stok, dan mencatat transaksi.
    """
    global total_pendapatan

    print("\n--- Penjualan Hasil Panen ---")

    nama_pembeli = input("Masukkan nama pembeli: ").strip()

    if not nama_pembeli:
        print("Nama pembeli tidak boleh kosong.")
        return

    tampilkan_data_panen()

    komoditas_dipilih = input(
        "Masukkan ID atau nama komoditas yang ingin dibeli: "
    ).strip().lower()

    item_terpilih = None

    for item in data_panen:
        if (
            item['id'].lower() == komoditas_dipilih
            or item['komoditas'].lower() == komoditas_dipilih
        ):
            item_terpilih = item
            break

    if not item_terpilih:
        print("Komoditas tidak ditemukan. Silakan masukkan ID atau nama yang benar.")
        return

    try:
        jumlah_beli = int(
            input(f"Masukkan jumlah {item_terpilih['komoditas']} (kg): ")
        )

        if jumlah_beli <= 0:
            print("Jumlah pembelian harus lebih dari 0.")
            return

    except ValueError:
        print("Input jumlah tidak valid. Harap masukkan angka.")
        return

    if jumlah_beli > item_terpilih['stok']:
        print(
            f"Stok {item_terpilih['komoditas']} tidak mencukupi. "
            f"Stok tersedia: {item_terpilih['stok']} kg."
        )
        return

    # Menghitung total harga sebelum diskon
    total_harga = jumlah_beli * item_terpilih['harga']

    # Menghitung diskon
    persentase_diskon, jumlah_diskon, total_setelah_diskon = hitung_diskon(
        total_harga
    )

    # Mengurangi stok
    item_terpilih['stok'] -= jumlah_beli

    # Menambahkan pendapatan berdasarkan harga setelah diskon
    total_pendapatan += total_setelah_diskon

    # Menyimpan transaksi
    transaksi = {
        'pembeli': nama_pembeli,
        'komoditas': item_terpilih['komoditas'],
        'jumlah': jumlah_beli,
        'harga_satuan': item_terpilih['harga'],
        'total_harga': total_harga,
        'diskon': persentase_diskon,
        'jumlah_diskon': jumlah_diskon,
        'total_bayar': total_setelah_diskon
    }

    riwayat_transaksi.append(transaksi)

    # Menampilkan hasil transaksi
    print(f"\nTransaksi berhasil untuk {nama_pembeli}!")
    print(f"Komoditas          : {transaksi['komoditas']}")
    print(f"Jumlah              : {transaksi['jumlah']} kg")
    print(f"Total harga         : Rp{transaksi['total_harga']:,.0f}")
    print(f"Diskon              : {transaksi['diskon']}%")
    print(f"Potongan harga      : Rp{transaksi['jumlah_diskon']:,.0f}")
    print(f"Total yang dibayar  : Rp{transaksi['total_bayar']:,.0f}")
    print(
        f"Stok {transaksi['komoditas']} tersisa: "
        f"{item_terpilih['stok']} kg"
    )
    print()


def tampilkan_riwayat():
    """Menampilkan seluruh riwayat transaksi penjualan."""
    print("\n--- Riwayat Transaksi ---")

    if not riwayat_transaksi:
        print("Belum ada transaksi yang tercatat.")
        return

    for i, transaksi in enumerate(riwayat_transaksi):
        print(f"Transaksi #{i + 1}:")
        print(f"  Pembeli          : {transaksi['pembeli']}")
        print(f"  Komoditas        : {transaksi['komoditas']}")
        print(f"  Jumlah           : {transaksi['jumlah']} kg")
        print(f"  Harga Satuan     : Rp{transaksi['harga_satuan']:,.0f}")
        print(f"  Total Harga      : Rp{transaksi['total_harga']:,.0f}")
        print(f"  Diskon            : {transaksi['diskon']}%")
        print(f"  Potongan Harga   : Rp{transaksi['jumlah_diskon']:,.0f}")
        print(f"  Total Pembayaran : Rp{transaksi['total_bayar']:,.0f}")
        print("------------------------")

    print()


def hitung_total_pendapatan():
    """Menampilkan total pendapatan dari seluruh transaksi."""
    print("\n--- Total Pendapatan ---")
    print(f"Total Pendapatan Saat Ini: Rp{total_pendapatan:,.0f}")
    print()


def main_menu():
    """Menampilkan menu utama program."""
    while True:
        print("--- Menu Utama Sistem Penjualan Hasil Panen ---")
        print("1. Lihat Data Panen")
        print("2. Lakukan Penjualan")
        print("3. Lihat Riwayat Transaksi")
        print("4. Lihat Total Pendapatan")
        print("5. Keluar")

        pilihan = input("Masukkan pilihan Anda (1-5): ").strip()

        if pilihan == '1':
            tampilkan_data_panen()

        elif pilihan == '2':
            jual_panen()

        elif pilihan == '3':
            tampilkan_riwayat()

        elif pilihan == '4':
            hitung_total_pendapatan()

        elif pilihan == '5':
            print("Terima kasih telah menggunakan Sistem Penjualan Hasil Panen.")
            break

        else:
            print("Pilihan tidak valid. Silakan coba lagi.")


# Memulai program
main_menu()
```
